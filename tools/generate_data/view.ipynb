```python
import numpy as np
import matplotlib.pyplot as plt

DATASET = 'dataset.npz'
```

```python
with np.load(DATASET, allow_pickle=True) as data:
    images = data['images']
    labels = data['labels']

print(f"Loaded {len(images)} images and {len(labels)} metadata rows.")
print(f"Image shape: {images.shape}, Labels shape: {labels.shape}")
```

```python
# Display the first image
plt.imshow(images[0], cmap='gray')
plt.title('First Image')
plt.axis('off')
plt.show()
```
